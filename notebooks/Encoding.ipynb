### Read and Clean Interaction

```python
interactions_df = pd.read_csv(os.path.join(path, "filtered_movie_ratings_merged.csv"))
interactions_df.info()
```

```python
interactions_df.dropna(inplace=True)
interactions_df.reset_index(inplace=True, drop=True)

interactions_df = interactions_df.rename(columns={'userId': 'user_id', 'tmdbId': 'item_id'})
interactions_df.drop(columns=['movieId'], inplace=True)

interactions_df["user_id"] = interactions_df["user_id"].astype("int64")
interactions_df["item_id"] = interactions_df["item_id"].astype("int64")
interactions_df["timestamp"] = interactions_df["timestamp"].astype("int64")
interactions_df["rating"] = interactions_df["rating"].astype("float32")

# per-user ordinal ordering -- needed by groupby_sequences(sort_col="timestamp")
# in every downstream notebook, so we bake it in once, here.
interactions_df = interactions_df.sort_values(by="timestamp")
interactions_df["timestamp"] = interactions_df.groupby("user_id").cumcount()

interactions_df.info()
```

### Read and Prepare item content features

```python
movie_df = pd.read_csv(os.path.join(path, "tmdb_movie_data_parsed_progressed_lemmaed.csv"))
movie_df.rename(columns={"id": "item_id"}, inplace=True)
movie_df.info()
```

```python
movie_df["genres"] = movie_df["genres"].apply(
    lambda x: literal_eval(x) if isinstance(x, str) else x
)
movie_df
```

```python
embeddding = SentenceTransformer("BAAI/bge-m3", trust_remote_code=True)
sentences = (movie_df["description"]).tolist()  # TODO: Experiment with different combination
```

```python
tokenizer = embeddding.tokenizer
token_count = []
for text in sentences:
    tokens = tokenizer.encode(text, truncation=False)
    token_count.append(len(tokens))
movie_df['tokenCount'] = token_count
# Optimizing the model for a speed boost
print("Default max_seq_length:", embeddding.max_seq_length)
embeddding.max_seq_length = max(token_count)
print("Updated max_seq_length:", embeddding.max_seq_length)
```

```python
embeddings = embeddding.encode(sentences)
movie_df['embeddings'] = list(embeddings)
movie_df.info()
```

```python
movie_df["keywords"].fillna("", inplace=True)
```

```python
tfidf = TfidfVectorizer(max_features=5000, min_df=2)   # min_df=2 drops keywords that appear in only 1 movie -- pure noise for similarity
tfidf_matrix = tfidf.fit_transform(movie_df["keywords"])
```

```python
svd = TruncatedSVD(n_components=128, random_state=42)
keyword_vecs = svd.fit_transform(tfidf_matrix)

# sklearn defaults to float64 -- keep everything float32 downstream
keyword_vecs = keyword_vecs.astype(np.float32)
movie_df["keyword_embedding"] = keyword_vecs.tolist()

print(svd.explained_variance_ratio_.sum())   # e.g. 0.6 means SVD retained 60% of the variance
```

```python
# Identify integer columns to normalize (excluding 'item_id')
numeric_cols = ['budget', 'revenue', 'runtime', 'year']

scaler = MinMaxScaler()
movie_df[numeric_cols] = scaler.fit_transform(movie_df[numeric_cols]).astype(np.float32)

print("Normalized integer columns in movie_df:")
print(movie_df[numeric_cols].head())
```

### Fit an Encoder

```python
encoder = LabelEncoder(
    [
        LabelEncodingRule("user_id"),
        LabelEncodingRule("item_id"),
    ]
)
encoder.rules[0].fit(interactions_df)
encoder.rules[1].fit(movie_df)

encoded_interactions = encoder.transform(interactions_df)
encoded_interactions
```

```python
item_features_encoded = encoder.rules[1].transform(movie_df)

item_features_encoded["genre_names"] = item_features_encoded["genres"]

genre_encoder = LabelEncoder(
    [
        SequenceEncodingRule("genres"),
    ]
)
genre_encoder.rules[0].fit(item_features_encoded)
item_features_encoded = genre_encoder.transform(item_features_encoded)

item_features_encoded["item_numerics"] = list(
    item_features_encoded[numeric_cols].values.astype(np.float32)
)
item_features_encoded
```

### Persist everything to Drive

```python
# Define paths
ENCODED_DIR = Path(path) / "encoded"
ENCODED_DIR.mkdir(parents=True, exist_ok=True)

ENCODER_PATH = ENCODED_DIR / "encoder"
INTERACTIONS_PATH = ENCODED_DIR / "encoded_interactions.parquet"
ITEM_FEATURES_PATH = ENCODED_DIR / "item_features_encoded.parquet"
```

```python
encoded_interactions.to_parquet(INTERACTIONS_PATH)

item_features_encoded[
    ["item_id", "title", "genres", "genre_names", "item_numerics", "embeddings", "keyword_embedding"]
].to_parquet(ITEM_FEATURES_PATH)

encoder.save(ENCODER_PATH)

print("Saved:")
print(" -", ENCODER_PATH)
print(" -", INTERACTIONS_PATH)
print(" -", ITEM_FEATURES_PATH)
```
