```python
from replay.preprocessing import LabelEncoder

ENCODED_DIR = Path(path) / "encoded"
ENCODER_PATH = ENCODED_DIR / "encoder"
INTERACTIONS_PATH = ENCODED_DIR / "encoded_interactions.parquet"
ITEM_FEATURES_PATH = ENCODED_DIR / "item_features_encoded.parquet"

# consumed later by ContentAwareEmbedder / FeaturesReader -- same file, kept under its
# original name so the rest of the notebook doesn't need to change
PATH_ENCODED_FEATURES = ITEM_FEATURES_PATH

encoder = LabelEncoder.load(ENCODER_PATH)
encoded_interactions = pd.read_parquet(INTERACTIONS_PATH)
item_features_encoded = pd.read_parquet(ITEM_FEATURES_PATH)
item_features_encoded.info()
```

### Train/val/test split

```python
from replay.splitters import LastNSplitter

splitter = LastNSplitter(
    N=1,
    divide_column="user_id",
    query_column="user_id",
    strategy="interactions",
    drop_cold_users=True,
    drop_cold_items=True
)

test_events, test_gt = splitter.split(encoded_interactions)
validation_events, validation_gt = splitter.split(test_events)
train_events = validation_events
```

```python
from replay.data.nn.utils import groupby_sequences


def bake_data(full_data):
    return groupby_sequences(events=full_data, groupby_col="user_id", sort_col="timestamp")
```

```python
train_events = bake_data(train_events)

validation_events = bake_data(validation_events)
validation_gt = bake_data(validation_gt)

test_events = bake_data(test_events)
test_gt = bake_data(test_gt)

train_events
```

```python
def add_gt_to_events(events_df, gt_df):
    gt_to_join = gt_df[["user_id", "item_id"]].rename(columns={"item_id": "ground_truth"})

    events_df = events_df.merge(gt_to_join, on="user_id", how="inner")
    return events_df

validation_events = add_gt_to_events(validation_events, validation_gt)
test_events = add_gt_to_events(test_events, test_gt)
```

```python
for df in (train_events, validation_events, test_events):  # and test_df / predict_df if applicable
    df["rating"] = df["rating"].apply(lambda seq: np.asarray(seq, dtype=np.float32))
```

```python
data_dir = Path("temp/data/")
data_dir.mkdir(parents=True, exist_ok=True)

TRAIN_PATH = data_dir / "train.parquet"
VAL_PATH = data_dir / "val.parquet"
PREDICT_PATH = data_dir / "test.parquet"
```

```python
train_events.to_parquet(TRAIN_PATH)
validation_events.to_parquet(VAL_PATH)
test_events.to_parquet(PREDICT_PATH)
```

### Traing

```python
from replay.data import FeatureHint, FeatureType
from replay.data.nn import TensorFeatureInfo, TensorSchema, TensorFeatureSource
from replay.data import FeatureSource

EMBEDDING_DIM = 64

NUM_UNIQUE_ITEMS = len(encoder.mapping["item_id"])

tensor_schema = TensorSchema(
    [
        TensorFeatureInfo(
            name="item_id",
            is_seq=True,
            padding_value=NUM_UNIQUE_ITEMS,
            cardinality=NUM_UNIQUE_ITEMS,
            embedding_dim=EMBEDDING_DIM,
            feature_type=FeatureType.CATEGORICAL,
            feature_hint=FeatureHint.ITEM_ID,
        ),
        TensorFeatureInfo(
            name="rating",
            is_seq=True,
            feature_type=FeatureType.NUMERICAL,
            tensor_dim=1,
            embedding_dim=EMBEDDING_DIM,  # must match item_id's, SumAggregator requires it
            feature_sources=[TensorFeatureSource(FeatureSource.INTERACTIONS, "rating")],
        ),
    ]
)
```

```python
from replay.nn.transform.template import make_default_sasrec_transforms
from replay.nn.transform import RenameTransform, CopyTransform, GroupTransform

item_id_name = tensor_schema.item_id_feature_name

custom_predict_transforms = [
    RenameTransform({f"{item_id_name}_mask": "padding_mask"}),
    CopyTransform({item_id_name: "seen_ids"}),          # <- survives the grouping below
    GroupTransform({"feature_tensors": tensor_schema.names}),
]

transforms = make_default_sasrec_transforms(tensor_schema)
# TODO: Delete if not needed
transforms["predict"] = custom_predict_transforms   # only override the predict split
```

```python
MAX_SEQ_LEN = 50

train_metadata = {
    "train": {
        "item_id": {"shape": MAX_SEQ_LEN + 1, "padding": tensor_schema["item_id"].padding_value},
        "rating": {"shape": MAX_SEQ_LEN + 1, "padding": 0.0},
    },
    "validate": {
        "item_id": {"shape": MAX_SEQ_LEN, "padding": tensor_schema["item_id"].padding_value},
        "rating": {"shape": MAX_SEQ_LEN, "padding": 0.0},
        "ground_truth": {"shape": 1, "padding": -1}
    },
}
```

```python
from replay.data.nn import ParquetModule

BATCH_SIZE = 32

parquet_module = ParquetModule(
    train_path=TRAIN_PATH,
    validate_path=VAL_PATH,
    batch_size=BATCH_SIZE,
    metadata=train_metadata,
    transforms=transforms,
)
```

```python
from replay.nn.sequential import SasRec

NUM_BLOCKS = 2
NUM_HEADS = 2
DROPOUT = 0.3

sasrec = SasRec.from_params(
    schema=tensor_schema,
    embedding_dim=EMBEDDING_DIM,
    max_sequence_length=MAX_SEQ_LEN,
    num_heads=NUM_HEADS,
    num_blocks=NUM_BLOCKS,
    dropout=DROPOUT,
)
```

```python
from replay.nn.lightning import LightningModule

model = LightningModule(sasrec)
```

```python
import torch
import lightning as L

torch.set_num_threads(1)
torch.set_num_interop_threads(1)

# L.seed_everything(42);
```

```python
from lightning.pytorch.callbacks import ModelCheckpoint
from lightning.pytorch.loggers import CSVLogger

from replay.nn.lightning.callback import ComputeMetricsCallback


checkpoint_callback = ModelCheckpoint(
    dirpath="sasrec/checkpoints",
    save_top_k=1,
    verbose=True,
    monitor="recall@10",
    mode="max",
)

validation_metrics_callback = ComputeMetricsCallback(
    metrics=["map", "ndcg", "recall"],
    ks=[1, 5, 10, 20],
    item_count=NUM_UNIQUE_ITEMS,
    # verbose=False,
)

csv_logger = CSVLogger(save_dir="sasrec/logs/train", name="SasRec-example")

trainer = L.Trainer(
    max_epochs=5,
    callbacks=[checkpoint_callback, validation_metrics_callback],
    logger=csv_logger,
)

trainer.fit(model, datamodule=parquet_module)
```

```python
best_model_path = checkpoint_callback.best_model_path
best_epoch = int(best_model_path.split("epoch=")[1].split("-")[0])

best_model_path
```

### Inference

```python
sasrec = SasRec.from_params(
    schema=tensor_schema,
    embedding_dim=EMBEDDING_DIM,
    max_sequence_length=MAX_SEQ_LEN,
    num_heads=NUM_HEADS,
    num_blocks=NUM_BLOCKS,
    dropout=DROPOUT,
    excluded_features=None,
)

best_model = LightningModule.load_from_checkpoint(best_model_path, model=sasrec)
best_model.eval();
```

```python
inference_metadata = {
    "predict": {
        "user_id": {},
        "item_id": {"shape": MAX_SEQ_LEN, "padding": tensor_schema["item_id"].padding_value},
        "rating": {"shape": MAX_SEQ_LEN, "padding": 0.0},
    }
}

parquet_module = ParquetModule(
    predict_path=PREDICT_PATH,
    batch_size=BATCH_SIZE,
    metadata=inference_metadata,
    transforms=transforms,
)
```

```python
from replay.nn.lightning.callback import PandasTopItemsCallback
from replay.nn.lightning.postprocessor import SeenItemsFilter

csv_logger = CSVLogger(save_dir="sasrec/logs/test", name="SasRec-example")
seen_filter = SeenItemsFilter(item_count=NUM_UNIQUE_ITEMS, seen_items_column="seen_ids")

TOPK = [1, 5, 10, 20]

pandas_prediction_callback = PandasTopItemsCallback(
    top_k=max(TOPK),
    query_column="user_id",
    item_column="item_id",
    rating_column="score",
    postprocessors=[seen_filter],
)

trainer = L.Trainer(callbacks=[pandas_prediction_callback], logger=csv_logger, inference_mode=True)
trainer.predict(best_model, datamodule=parquet_module, return_predictions=False)

pandas_res = pandas_prediction_callback.get_result()
```

```python
import torch.nn.functional as F

item_embeddings = best_model.model.body.embedder.get_item_weights()

def explain_by_similarity_and_rating(user_history_item_ids, user_ratings, recommended_item_id, top_n=3):
    rec_vec = item_embeddings[recommended_item_id]
    hist_vecs = item_embeddings[user_history_item_ids]
    sims = F.cosine_similarity(rec_vec.unsqueeze(0), hist_vecs)
    top_idx = sims.topk(min(top_n, len(user_history_item_ids))).indices
    return [
        {
            "item_id": user_history_item_ids[i],
            "similarity": sims[i].item(),
            "your_rating": user_ratings[i],
        }
        for i in top_idx
    ]
```

```python
def format_explanation_collab(explanation, rec_name):
    def bucket(sim):
        if sim > 0.8: return "very similar to"
        if sim > 0.6: return "similar to"
        if sim > 0.4: return "related to"
        return None

    lines = []
    for exp in explanation:
        if bucket(exp["similarity"]) is None:
            continue
        title = movie_names.get(item_id_decoder[exp["item_id"]], "an unknown title")
        lines.append(f'{title} (your rating: {exp["your_rating"]})')

    if not lines:
        return f'"{rec_name}" is broadly similar to your recent viewing.'
    return f'Because you enjoyed {", ".join(lines)}, we think you\'ll like "{rec_name}".'
```

```python
item_id_decoder = encoder.inverse_mapping["item_id"]

movie_names = {
    item_id_decoder.get(int(row.item_id), int(row.item_id)): row.title
    for row in item_features_encoded.itertuples()
}
```

```python
user_histories = (
    test_events.set_index("user_id")[["item_id", "rating"]]
    .to_dict(orient="index")
)
# user_histories
```

```python
from tqdm import tqdm

def add_explanations_collab(pandas_res, user_histories):
    explanations = []
    for _, row in tqdm(pandas_res.iterrows()):
        user_id = row["user_id"]
        rec_item_id_raw = int(row["item_id"])          # already raw, post inverse_transform
        history = user_histories.get(user_id)
        if history is None:
            explanations.append(None)
            continue

        recommended_item_id_encoded = encoder.mapping["item_id"].get(rec_item_id_raw)
        if recommended_item_id_encoded is None:
            explanations.append(None)
            continue

        explanation = explain_by_similarity_and_rating(
            history["item_id"], history["rating"], recommended_item_id_encoded
        )
        rec_name = movie_names.get(rec_item_id_raw, "Unknown title")
        explanations.append(format_explanation_collab(explanation, rec_name))

    pandas_res = pandas_res.copy()
    pandas_res["explanation"] = explanations
    return pandas_res
```
